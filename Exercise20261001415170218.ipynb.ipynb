1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。

將十進位小數轉成二進位的方法：

反覆乘以 2，取整數部分

| 步驟  | 計算            | 整數部分 |
| --- | ------------- | ---- |
| 1   | 0.1 × 2 = 0.2 | 0    |
| 2   | 0.2 × 2 = 0.4 | 0    |
| 3   | 0.4 × 2 = 0.8 | 0    |
| 4   | 0.8 × 2 = 1.6 | 1    |
| 5   | 0.6 × 2 = 1.2 | 1    |
| 6   | 0.2 × 2 = 0.4 | 0    |
| 7   | 0.4 × 2 = 0.8 | 0    |
| 8   | 0.8 × 2 = 1.6 | 1    |
| ... | ...           | ...  |

因此：

0.1₁₀
=
0.00011001100110011001100110011...₂

可寫成：

0.1 =
0.0001(1001)₂

其中：

1001

會無限循環

    現象

就像：

1/3 = 0.333333333...

在十進位中無法有限表示。

同樣地：

1/10 = 0.1

在二進位中也無法有限表示。

因此電腦無法把 0.1 完全存進有限長度的記憶體中。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙

Python 的 float 採用 IEEE 754 雙精度浮點數（binary64）格式。

一個浮點數共有：

1 位符號位

11 位指數位

52 位尾數位

實際有效精度約為：

53 個二進位位元
≈ 15~17 位十進位有效數字

因此：

0.000110011001100110011001...

只能保留前面約 53 位，

剩下的部分必須進行捨入。

IEEE 754 預設使用：

        Round Half to Even

中文通常稱為：

四捨六入五成雙

例如：

2.5 -> 2
3.5 -> 4

1.25 -> 1.2
1.35 -> 1.4

這種做法可降低長期累積誤差。
