習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png

import random

r = random.random()

print(f"r = {r:.16f}")

x = r
binary = ""

count = 0
while count < 20:
    x *= 2

    if x >= 1:
        binary += "1"
        x -= 1
    else:
        binary += "0"

    count += 1

print(f"r ≈ 0.{binary}₂")

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

並查閱官方文件，了解：

- `bit_length()`
- `sys.getsizeof()`
- `PyLongObject`

三者之間的關係。

習題二：輸入一個正整數 \(n\)，其中$1 \le n \le 10^{68}$，請將 \(n\) 轉換為中文大寫數字表示法。需支援下列數字單位：萬、億、兆、京、垓、秭、穰、溝、澗、正、載、極、恆河沙、阿僧祇、那由他、不可思議、無量大數。比如輸入31415926，輸出參仟壹佰肆拾壹萬伍仟玖佰貳拾陸。

提示：思考利用字典，建立「數字」與「中文大寫數字」的對應關係。

digits = {
    '0': '零',
    '1': '壹',
    '2': '貳',
    '3': '參',
    '4': '肆',
    '5': '伍',
    '6': '陸',
    '7': '柒',
    '8': '捌',
    '9': '玖'
}

small_units = ['', '拾', '佰', '仟']

big_units = [
    '', '萬', '億', '兆', '京', '垓', '秭', '穰',
    '溝', '澗', '正', '載', '極',
    '恆河沙', '阿僧祇', '那由他',
    '不可思議', '無量大數'
]


def four_digit_to_chinese(num_str):
    """將 0~9999 的字串轉成中文大寫"""
    
    num_str = num_str.zfill(4)
    result = ""

    zero_flag = False

    for i in range(4):
        digit = num_str[i]
        pos = 3 - i

        if digit != '0':
            if zero_flag:
                result += '零'
                zero_flag = False

            result += digits[digit] + small_units[pos]

        else:
            if result:
                zero_flag = True

    return result


def to_chinese(n):
    if n == 0:
        return "零"

    s = str(n)

    groups = []

    while s:
        groups.append(s[-4:])
        s = s[:-4]

    result = ""

    for i in range(len(groups) - 1, -1, -1):

        part = four_digit_to_chinese(groups[i])

        if part:
            result += part + big_units[i]

    return result


n = int(input("請輸入整數："))
if not (1 <= n <= 10**68):
    print("輸入超出範圍")
else:    
    print(to_chinese(n))
